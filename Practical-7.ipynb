```python
import numpy as np

# Create NumPy array
arr = np.array([10, 20, 30, 40, 50])

print("Original Array:")
print(arr)

# Addition
print("\nAddition:")
print(arr + 5)

# Subtraction
print("\nSubtraction:")
print(arr - 5)

# Multiplication
print("\nMultiplication:")
print(arr * 5)

# Division
print("\nDivision:")
print(arr / 5)
```

```python
import numpy as np

# Create two arrays
a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])

print("Array A:")
print(a)

print("\nArray B:")
print(b)

# Addition
print("\nAddition:")
print(a + b)

# Subtraction
print("\nSubtraction:")
print(a - b)

# Multiplication
print("\nMultiplication:")
print(a * b)

# Division
print("\nDivision:")
print(a / b)
```

```python
import numpy as np

# Create 2D array
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Original Array:")
print(arr)

# Scalar broadcasting
print("\nArray + 5:")
print(arr + 5)

print("\nArray - 5:")
print(arr - 5)

print("\nArray * 2:")
print(arr * 2)

print("\nArray / 2:")
print(arr / 2)
```

```python
import numpy as np

# Create an array
arr = np.array([1, 4, 9, 16, 25])

print("Original Array:")
print(arr)

# Square root
sqrt_result = np.sqrt(arr)

print("\nSquare Root:")
print(sqrt_result)

# Logarithm
log_result = np.log(arr)

print("\nNatural Logarithm:")
print(log_result)

# Sine
sin_result = np.sin(arr)

print("\nSine:")
print(sin_result)
```

```python
import numpy as np

# Create array
arr = np.array([1, 2, 3, 4, 5])

print("Original Array:")
print(arr)

# Calculate sine
sin_result = np.sin(arr)

print("\nSin(Array):")
print(sin_result)

# Multiply array with sin(array)
result = arr * np.sin(arr)

print("\nArray * Sin(Array):")
print(result)
```
