### Create the function that will decide if provided numbers consitute pythagorean triples.

```python
def is_pythagorean(a, b, c):
    sides = sorted([a, b, c])
    
    if sides[0]**2 + sides[1]**2 == sides[2]**2:
        return True
    else:
        return False


a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if is_pythagorean(a, b, c):
    print("The numbers are Pythagorean triples.")
else:
    print("The numbers are not Pythagorean triples.")
```

### Create the function that will take list of any length from user and return the last value in the list

```python
def last_value(*args):
    return args[-1]


numbers = input("Enter numbers separated by spaces: ").split()
numbers = [int(x) for x in numbers]

print("Last value:", last_value(*numbers))
```

### Create function for grading system and handle invalid marks, marks would be in range 0 to 100

```python
def grading_system(marks):
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"


try:
    marks = float(input("Enter marks (0-100): "))
    grade = grading_system(marks)
    print("Grade:", grade)

except ValueError as e:
    print("Invalid marks:", e)
```
